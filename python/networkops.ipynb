```python
url = "https://dummyjson.com/products"
# url = "https://dummyjson.com/carts"
# url = "https://dummyjson.com/users"
query_params = {"limit": 10, "skip": 5}

response = requests.get(url, params=query_params)
print(f"status code: {response.status_code}")

response.raise_for_status()      # fail loudly on 4xx/5xx instead of crashing in .json()
data = response.json()
print(data)
```

```python
import requests

login = requests.post(
    "https://dummyjson.com/auth/login",
    json={"username": "emilys", "password": "emilyspass", "expiresInMins": 30},
)
print(f"status: {login.raise_for_status()}")
print(f"response: {login.json()}")      
token = login.json()["accessToken"]

me = requests.get(
    "https://dummyjson.com/auth/me",
    headers={"Authorization": f"Bearer {token}"},
)
print(me.json()["email"])
```
