`
MRO (Method Resolution Order) is the order Python follows when looking up a method or attribute on an object. It checks the object’s class first, then its parents, and so on, and uses the first match it finds.
`


```python
class A:
    def hello(self): print("A")

class B(A):
    def hello(self): print("B")

class C(A):
    def hello(self): print("C")

class D(B, C): pass

D().hello()       # prints "B"
print(D.mro())    # [D, B, C, A, object]
```

## D doesn’t define hello, so Python moves through the MRO: it checks B next, finds hello, and stops.

## The three rules the order always respects:

1.A child comes before its parents.
2.Parents keep the left-to-right order from the class definition.
3.Each class appears only once, so a shared ancestor like A is placed after all the classes that inherit from it.
