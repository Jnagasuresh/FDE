In Python, **type casting** (or type conversion) is the process of converting a value from one data type to another.

Python supports two kinds of casting: **implicit** (done automatically by the interpreter) and **explicit** (done manually using built-in constructor functions).

---

### 1. Implicit Type Conversion

Python automatically converts one data type into another to prevent data loss without requiring user intervention.

```python
x = 10    # int
y = 2.5   # float

total = x + y  # Python automatically casts 'x' to float
print(total)        # 12.5
print(type(total))  # <class 'float'>

```

---

### 2. Explicit Type Conversion

Explicit casting is performed manually using constructor functions.

#### Common Built-in Conversion Functions

| Target Type | Function | Example | Result |
| --- | --- | --- | --- |
| **Integer** | `int(x, base=10)` | `int("42")` / `int(3.99)` | `42` / `3` *(truncates towards zero)* |
| **Float** | `float(x)` | `float("3.14")` / `float(7)` | `3.14` / `7.0` |
| **String** | `str(x)` | `str(100)` / `str([1, 2])` | `'100'` / `'[1, 2]'` |
| **Boolean** | `bool(x)` | `bool(0)` / `bool("text")` | `False` / `True` |
| **List** | `list(x)` | `list("cat")` / `list((1, 2))` | `['c', 'a', 't']` / `[1, 2]` |
| **Tuple** | `tuple(x)` | `tuple([1, 2, 3])` | `(1, 2, 3)` |
| **Set** | `set(x)` | `set([1, 2, 2, 3])` | `{1, 2, 3}` *(removes duplicates)* |
| **Dictionary** | `dict(x)` | `dict([('a', 1), ('b', 2)])` | `{'a': 1, 'b': 2}` |

---

### Key Nuances & Pitfalls

#### A. Truncation vs. Rounding

Casting a float to an `int` **truncates** the decimal portion; it does not round:

```python
int(4.9)   # 4
int(-4.9)  # -4
round(4.9) # 5

```

#### B. String to Int with Non-Base-10

`int()` accepts a second argument specifying the radix/base:

```python
int("1010", base=2)  # 10
int("1a", base=16)   # 26

```

#### C. Two-Step String Float Conversion

You cannot pass a decimal string directly into `int()`:

```python
int("3.14")  # Raises ValueError: invalid literal for int() with base 10: '3.14'

# Correct approach:
int(float("3.14"))  # 3

```

#### D. Truthiness in `bool()`

`bool()` evaluates the "truth value" of an object rather than parsing words:

```python
bool("")       # False (empty string)
bool("False")  # True  (non-empty string is always True)
bool(0)        # False
bool([])       # False (empty collection)

```

### Custom Type Casting in Classes

You can define how your custom classes respond to casting functions by implementing Python's magic methods:

```python
class Temperature:
    def __init__(self, celsius):
        self.celsius = celsius

    def __int__(self):
        return int(self.celsius)

    def __float__(self):
        return float(self.celsius)

    def __str__(self):
        return f"{self.celsius}°C"

temp = Temperature(21.75)
print(int(temp))    # 21
print(float(temp))  # 21.75
print(str(temp))    # 21.75°C

```
