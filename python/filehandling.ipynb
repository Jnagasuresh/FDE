File Handling

```python
f = open("filehandling.txt","r")
content = f.read()
print(content)

f.close()
```

```python
# file with context manager
with open("filehandling.txt","r") as f:
    content =f.read()
    print(content)
```

```python
with open("inspection1.txt","w") as f:
    f.write("content starts here")
```

File Modes:
Mode: r,w,a,r+ (read + write), w+ (write + read), a+ (append + read)

file -> reading ---> read() -> full content
                ---> readline()
                ---- readlines()
        writing ---> write()
                ---  writeliens(line, line2)
        seek   --->  seek()
               --->  seek(0)
