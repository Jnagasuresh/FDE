Identity: `is` vs `==`
== checks whether values are equal; 
`is` checks whether two names point to the same object in memory.

Rule of thumb: use `is` only for `None`, `True`, `False`; 
use `==` for everything else.

````python
a = [1, 2]
b = [1, 2]
print(a == b)   # True  (same contents)
print(a is b)   # False (different objects)

x = None
if x is None:   # correct way to check for None
    ...
````

3. Membership: `in` and `not in`

Membership on a `set` or `dict` is very fast (O(1)); on a `list` it’s a linear scan.

````python
print("a" in "banana")          # True
print(3 in [1, 2, 3])           # True
print("key" in {"key": 1})      # True (checks dict keys)
print(5 not in range(3))        # True
````

unpacking `*` and `**`

In Python, the unpacking operators are the single asterisk (`*`) and double asterisk (`**`). They allow you to extract individual elements from iterables and mappings respectively.

---

### 1. The Iterable Unpacking Operator (`*`)

The `*` operator works on any iterable (lists, tuples, sets, strings, generators).

#### A. Function Calls (Passing Arguments)

Unpacks elements of an iterable as separate positional arguments.

```python
def add(a, b, c):
    return a + b + c

nums = [1, 2, 3]
print(add(*nums))  # Equivalent to add(1, 2, 3) -> 6

```

#### B. Function Definitions (`*args`)

Collects an arbitrary number of positional arguments into a single tuple.

```python
def summarize(*args):
    print(type(args))  # <class 'tuple'>
    return sum(args)

summarize(10, 20, 30)  # 60

```

#### C. Merging and Creating Iterables (PEP 448)

Combines lists, tuples, or sets cleanly without explicit concatenation methods.

```python
list_a = [1, 2]
list_b = [3, 4]

merged = [*list_a, *list_b, 5]  # [1, 2, 3, 4, 5]

```

#### D. Extended Iterable Unpacking (Assignment / Slicing)

Captures remaining elements during assignment into a list.

```python
first, *middle, last = [10, 20, 30, 40, 50]
# first  = 10
# middle = [20, 30, 40]
# last   = 50

head, *tail = [1, 2, 3]
# head = 1, tail = [2, 3]

```

---

### 2. The Dictionary Unpacking Operator (`**`)

The `**` operator works on dictionary-like mappings (keys must typically be strings matching target parameter names).

#### A. Function Calls (Passing Keyword Arguments)

Unpacks key-value pairs into named keyword arguments.

```python
def configure(host, port, debug=False):
    print(f"Connecting to {host}:{port} (debug={debug})")

config = {"host": "localhost", "port": 8080, "debug": True}
configure(**config)  # Equivalent to configure(host="localhost", port=8080, debug=True)

```

#### B. Function Definitions (`**kwargs`)

Collects arbitrary keyword arguments into a standard dictionary.

```python
def log_metadata(**kwargs):
    print(type(kwargs))  # <class 'dict'>
    for key, value in kwargs.items():
        print(f"{key}: {value}")

log_metadata(user="alice", role="admin", attempts=3)

```

#### C. Merging Dictionaries

Combines multiple dictionaries into a new dictionary. Keys on the right overwrite keys on the left if duplicates exist.

```python
defaults = {"theme": "light", "notifications": True}
user_settings = {"theme": "dark"}

profile = {**defaults, **user_settings, "lang": "en"}
# {'theme': 'dark', 'notifications': True, 'lang': 'en'}

```

*(Note: In Python 3.9+, dictionary merging can also be written using the `|` union operator: `defaults | user_settings`.)*

---

### Quick Reference Summary

| Syntax | Target Type | Context | Behavior |
| --- | --- | --- | --- |
| `*iterable` | Lists, tuples, sets, generators | Function call | Passes items as positional arguments |
| `*name` | Any | Function signature | Collects extra positional arguments into a `tuple` |
| `*iterable` | Lists, tuples, sets | Data structure literal | Inlines items into the new sequence |
| `*rest` | Sequences | Variable assignment | Collects leftover items into a `list` |
| `**dict` | Dictionaries / mappings | Function call | Passes pairs as keyword arguments (`key=value`) |
| `**name` | Any | Function signature | Collects extra keyword arguments into a `dict` |
| `**dict` | Dictionaries / mappings | Dict literal | Inlines key-value pairs into a new dictionary |

In Python, matching items to variables by position is known as **positional unpacking** (or **sequence unpacking** / **tuple unpacking**).

Here are the primary ways it is used across the language:

---

### 1. Basic Sequence Unpacking

The number of variables on the left must exactly match the number of elements in the sequence on the right.

```python
point = (10, 20, 30)
x, y, z = point

print(x)  # 10
print(y)  # 20
print(z)  # 30

```

> **Note:** If the counts do not match, Python raises a `ValueError: too many values to unpack` or `not enough values to unpack`.

---

### 2. Extended Positional Unpacking (Using `*`)

Introduced in [PEP 3132](https://peps.python.org/pep-3132/), the starred variable captures any excess positional elements into a `list`.

```python
numbers = [1, 2, 3, 4, 5]

# Capture the start and end, pack the middle
first, *middle, last = numbers
# first = 1, middle = [2, 3, 4], last = 5

# Capture head and tail
head, *tail = numbers
# head = 1, tail = [2, 3, 4, 5]

```

---

### 3. Loop Iteration Unpacking

Unpacking frequently occurs directly in `for` loops when iterating over pairs or tuples:

```python
pairs = [(1, 'a'), (2, 'b'), (3, 'c')]

for index, letter in pairs:
    print(f"Index: {index}, Letter: {letter}")

```

This is also how `enumerate()` and `dict.items()` work:

```python
for idx, item in enumerate(['apple', 'banana']):
    ...

for key, value in {'a': 1, 'b': 2}.items():
    ...

```

---

### 4. Nested Positional Unpacking

Target variables can mirror the nested structure of the data:

```python
record = ("Alice", (2026, 10, 24), "Admin")
name, (year, month, day), role = record

print(name)   # "Alice"
print(year)   # 2026

```

---

### 5. Structural Pattern Matching (`match` / `case`)

In Python 3.10+, positional matching can be handled via sequence patterns:

```python
point = (0, 5)

match point:
    case (0, 0):
        print("Origin")
    case (0, y):
        print(f"On Y-axis at {y}")
    case (x, 0):
        print(f"On X-axis at {x}")
    case (x, y):
        print(f"Point at ({x}, {y})")

```

````python
# The simplest form matches items to variables by position:

point = (3, 5)
x, y = point            # x=3, y=5
print(x)
print(y)

name, age = ["Ravi", 25]
print(name)
print(age)
a, b, c = "abc"         # works on strings too
print(a)
print(b)
print(c)
print("----")
a, b = b, a             # swap: the right side is built as a tuple first
print(a)
print(b)

print('**************************')
# The count must match exactly, otherwise you get an error:
# a, b = [1, 2, 3]   # ValueError: too many values to unpack

print('Nested structures unpack too:')
# Nested structures unpack too:
data = ("Ravi", (25, "Vijayawada"))
name, (age, city) = data
print(name)
print(age)
print(city)
````

````python
#2. Extended unpacking with `*`` in assignment

# A starred variable collects “everything else” into a list:

first, *rest = [1, 2, 3, 4]       # first=1, rest=[2, 3, 4]
print(first)
print(*rest)
*start, last = [1, 2, 3, 4]       # start=[1, 2, 3], last=4
first, *middle, last = [1, 2, 3, 4, 5]   # middle=[2, 3, 4]
print(first)
print(*middle)
print(last)

head, *tail = [1]                 # head=1, tail=[]   (empty is fine)
print(head)
print(*tail)
````

3. Unpacking into function calls

* spreads an iterable into positional arguments:

````python
def add(a, b, c):
    return a + b + c

nums = [1, 2, 3]
add(*nums)          # same as add(1, 2, 3)

print(*nums)              # 1 2 3
print(*nums, sep=", ")    # 1, 2, 3
print(*"hello")           # h e l l o
````

`**` spreads a dictionary into keyword arguments, where keys must match parameter names:

````python
def greet(name, age, greeting):
    print(f"{greeting}, {name} is {age}!")

info = {"name": "Ravi", "greeting": "Hello","age":"40"}
greet(**info)       # same as greet(name="Ravi", greeting="Hello")
````

4. Packing in function definitions: `*args` and `**kwargs`

In a function definition, the stars do the reverse: they collect extra arguments.

````python
def show(*args, **kwargs):
    print("args:", args)       # a tuple
    print("kwargs:", kwargs)   # a dict

show(1, 2, 3, color="red", size=10)
# args: (1, 2, 3)
# kwargs: {'color': 'red', 'size': 10}
````
