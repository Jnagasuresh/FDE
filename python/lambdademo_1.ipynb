## Limitations and style

A lambda can hold only one expression: no statements, no assignments (except the := walrus operator), no loops, and no type hints. It also shows up as <lambda> in tracebacks, which makes debugging harder. PEP 8 recommends against assigning a lambda to a name (f = lambda x: ...); use def instead. Lambdas are best for short, throwaway functions passed directly as arguments, like the key= in sorting.

```python
from functools import reduce

nums = [1, 2, 3, 4, 5]
list(map(lambda x: x * 2, nums))          # [2, 4, 6, 8, 10]
list(filter(lambda x: x % 2 == 0, nums))  # [2, 4]
reduce(lambda acc, x: acc * x, nums)      # 120
```

```python
ops = {
    "add": lambda a, b: a + b,
    "sub": lambda a, b: a - b,
    "mul": lambda a, b: a * b,
}
ops["mul"](3, 4)   # 12
```

# A classic pitfall: lambdas in a loop
### Lambdas capture variables by reference, not by value, so all of these see the final i

```python
funcs = [lambda: i for i in range(3)]
[f() for f in funcs]          # [2, 2, 2]  — surprising
```
