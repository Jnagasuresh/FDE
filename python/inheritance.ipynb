```python
class Engine:
    def start(self):
        print("Engine started")

class Car(Engine):          # Car IS an Engine?
    def drive(self):
        self.start()
        print("Driving")
```

Has a: Composition

```python
class Car:
    def __init__(self, engine):
        self.engine = engine    # Car HAS an Engine

    def drive(self):
        self.engine.start()
        print("Driving")

car = Car(Engine())
car.drive()
```

# Inheritance vs Composition
## Key difference

`
With inheritance, the relationship is fixed when you write the class: Car(Engine) is permanent. With composition, it’s set at runtime, so you can pass in an ElectricEngine or a mock engine for testing without changing Car.

Inheritance couples the child tightly to the parent’s internals. A change in the parent can silently break children (the “fragile base class” problem). Composition only depends on the public methods of the object it holds.

Inheritance gives you the parent’s whole interface for free, which is convenient but can expose methods that don’t belong. Composition makes you write the delegating methods yourself, which is more typing but keeps the public interface deliberate.
`
## When to use which
`
Use inheritance when the “is-a” relationship is genuine and stable: a Square is a Shape, a ValueError is an Exception. It’s also the natural fit for implementing an ABC or framework base class, like the interfaces from the previous question.

Use composition when one object uses another, or when you want to swap behavior. The common design advice “favor composition over inheritance” exists because composition tends to produce code that’s easier to test and change, and it sidesteps the MRO complications of deep or multiple inheritance.

In practice, most well-designed code uses both: shallow inheritance for genuine type relationships, and composition for assembling behavior.
`
